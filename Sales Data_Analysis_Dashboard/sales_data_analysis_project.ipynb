```python
import numpy as np
import pandas as pd

print(np.__version__)
print(pd.__version__)
```

## uplaod csv files

```python
customers=pd.read_csv('customers.csv')
customers.head(5)
```

```python
orders=pd.read_csv('orders.csv')
orders.head(5)
```

```python
products=pd.read_csv('products.csv')
products.head(5)
```

```python
returns=pd.read_csv("returns.csv")
returns.head(5)
```

```python
sales=pd.read_csv('sales.csv')
sales.head(5)
```

## clean dtasets

```python
customers.dropna(inplace=True)
```

```python
orders.dropna(inplace=True)
```

```python
products.dropna(inplace=True)
```

```python
returns.dropna(inplace=True)
```

```python
sales.dropna(inplace=True)
```

```python
print(type(customers))
print(type(orders))
print(type(products))
print(type(sales))
print(type(returns))
```

## Merge datasets

```python
final_data=pd.merge(
    customers,
    orders,
    on='CustomerID'
)
```

```python
final_data=pd.merge(
    products,
    returns,
    on='ProductID'
)
```

```python
final_data=pd.merge(
    final_data,
    sales,
    on='OrderID'
)
final_data.head(5)
```
