```python
import numpy as np
import pandas as pd
```

### Employee csv files

```python
Employees=pd.read_csv('employees.csv')
Employees.head(3)
```

### Attendance csv files

```python
attendance=pd.read_csv('attendance.csv')
attendance.head(3)
```

### Salary csv file

```python
salary=pd.read_csv('salary.csv')
salary.head(3)
```

## Clean missing values

```python
Employees = Employees.dropna()
```

```python
attendance=attendance.dropna()
```

```python
salary=salary.dropna()
```

```python
print(type(Employees))
print(type(attendance))
Employees = pd.read_csv("employees.csv")
```

## Marge datasets

```python
Employees_attendance = pd.merge(
    Employees,
    attendance,
    on='EmployeeID'
)
```

```python
Final_data=pd.merge(
    Employees_attendance,
    salary,
    on='EmployeeID'
)
```

### Calculate present, absent, attendance percentage.

```python
Final_data['AttendancePercentage'] = (Final_data['PresentDays'] / Final_data['WorkingDays']) * 100
```

```python
Final_data[['EmployeeID', 'PresentDays', 'WorkingDays', 'AttendancePercentage']].head()
```

```python
Final_data.head()
```

## Calculate Salary

```python
#  daily salry
Final_data['daily_salary'] = Final_data['NetSalary'] / Final_data['WorkingDays']
```

```python
# final salary
Final_data['Final_salary'] = Final_data['daily_salary']*Final_data['PresentDays']
```

```python
Final_data.head()
```

```python
# Max salary

highest_salary=Final_data['Final_salary'].max()
highest_salary
```

```python
# lowest salary
lowest_salary=Final_data['Final_salary'].min()
lowest_salary
```

```python
# average salary

average_salary=Final_data['Final_salary'].mean()
average_salary
```

### Department-wise Salary

```python
Final_data.groupby('Department')['Final_salary'].mean()
```

```python
Final_data[Final_data['AttendancePercentage']<80.0].head(10)
```
